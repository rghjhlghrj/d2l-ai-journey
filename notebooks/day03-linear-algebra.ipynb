# Day 3 打卡：线性代数（上）—— 标量、向量、矩阵

来源：[2.3 线性代数](https://zh.d2l.ai/chapter_preliminaries/linear-algebra.html)（前半部分）

**要点摘录（读书笔记）**：
1. 标量=单个数；向量=一列数（有长度/维度）；矩阵=二维表格（行×列）
2. 矩阵的转置 A.T：行变列、列变行；方阵 A.T == A 则是对称矩阵
3. 深度学习的起点就是：用矩阵表示数据，用矩阵运算表示变换

```python
# 【1】标量：就是单独一个数，如温度
import torch

x = torch.tensor(3.0)
y = torch.tensor(2.0)

x + y, x * y, x / y, x**y
```

```python
# 【2】向量：一列数，arange(4) 生成 [0,1,2,3]；x[3] 按下标取元素
x = torch.arange(4)
x
```

```python
# 【3】向量的长度（元素个数）：len()；一维张量的形状只有一行 (4,)
x[3]
```

```python
# 【4】矩阵：二维张量，reshape(5,4) 把15个数排成5行4列
len(x)
```

```python
# 【5】A.T 是转置：行变列、列变行
x.shape
```

```python
# 【6】对称矩阵：A == A.T 说明转置后和原来一样
A = torch.arange(20).reshape(5, 4)
A
```

```python
# 【7】矩阵也是张量，len() 返回行数，形状是 (行, 列)
A.T
```
