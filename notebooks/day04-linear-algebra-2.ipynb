来源：[2.3 线性代数](https://zh.d2l.ai/chapter_preliminaries/linear-algebra.html)（后半部分）

**要点摘录（读书笔记）**：
1. 矩阵-向量积 Ax：把矩阵看成一堆行向量的加权组合，`torch.mv(A, x)`
2. 矩阵乘法 AB：C 的第 i 行第 j 列 = A 第 i 行与 B 第 j 列的点积，`torch.mm(A, B)`；注意≠逐元素乘法 A*B
3. 范数衡量向量'多大'：L2范数=平方和开根号（最常用），L1范数=绝对值求和（对离群点不敏感）
4. 解答 Day 3 的疑问：矩阵和向量相乘（mv）不是逐元素乘，而是行×列的组合求和

```python
import torch
# 【环境准备】复现原笔记本前半部分定义的变量
x = torch.arange(4, dtype=torch.float32)
A = torch.arange(20, dtype=torch.float32).reshape(5, 4)
B = A.clone()
u = torch.tensor([3.0, -4.0])
v = torch.ones_like(u)
```

```python
# 【1】回顾点积：两个向量对应位置相乘再求和
y = torch.ones(4, dtype = torch.float32)
x, y, torch.dot(x, y)
```

```python
# 【2】矩阵-向量积 Ax：A的每一行与x做点积，得到一个向量（解答Day3疑问：不是逐元素乘）
torch.sum(x * y)
```

```python
# 【3】矩阵-矩阵乘法 AB：形状 (2,3)×(3,4)→(2,4)，第i行第j列=A第i行与B第j列的点积
A.shape, x.shape, torch.mv(A, x)
```

```python
# 【4】L2范数：所有元素平方和开根号，torch.norm(u)，衡量向量长度
B = torch.ones(4, 3)
torch.mm(A, B)
```

```python
# 【5】L1范数：绝对值之和，torch.abs(u).sum()，比L2对异常值不那么敏感
u = torch.tensor([3.0, -4.0])
torch.norm(u)
```

```python
# 【6】矩阵的弗罗贝尼乌斯范数：把矩阵拉平后算L2范数
torch.abs(u).sum()
```
