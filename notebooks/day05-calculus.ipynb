# Day 5 打卡：微积分（导数、偏导数、梯度）

来源：[2.4 微积分](https://zh.d2l.ai/chapter_preliminaries/calculus.html)

**要点摘录（读书笔记）**：
1. 导数：函数在某点的瞬时变化率，几何上是曲线切线的斜率
2. 数值近似：用 (f(x+h)-f(x))/h 当 h 极小时近似导数
3. 梯度：多元函数对每个变量分别求偏导组成的向量，指向函数上升最快的方向；训练就是反着梯度走（梯度下降）

```python
# 【1】导入：绘制函数 f(x)=3x^2-4x，直观看切线
%matplotlib inline
import numpy as np
from matplotlib_inline import backend_inline
from d2l import torch as d2l


def f(x):
    return 3 * x ** 2 - 4 * x
```

```python
# 【2】数值近似导数：h→0 时 (f(x+h)-f(x))/h 收敛到真实导数(在x=1处为2)
def numerical_lim(f, x, h):
    return (f(x + h) - f(x)) / h

h = 0.1
for i in range(5):
    print(f'h={h:.5f}, numerical limit={numerical_lim(f, 1, h):.5f}')
    h *= 0.1
```

```python
# 【3】绘制 f(x) 与 x=1 处的切线，验证斜率约等于导数
def use_svg_display():  #@save
    """使用svg格式在Jupyter中显示绘图"""
    backend_inline.set_matplotlib_formats('svg')
```

```python
# 【4】偏导数：对多元函数 f=2x^2+3y^2 分别对 x、y 求偏导
def set_figsize(figsize=(3.5, 2.5)):  #@save
    """设置matplotlib的图表大小"""
    use_svg_display()
    d2l.plt.rcParams['figure.figsize'] = figsize
```

```python
# 【5】梯度：把所有偏导数拼成一个向量，torch.tensor 手动演示梯度方向
#@save
def set_axes(axes, xlabel, ylabel, xlim, ylim, xscale, yscale, legend):
    """设置matplotlib的轴"""
    axes.set_xlabel(xlabel)
    axes.set_ylabel(ylabel)
    axes.set_xscale(xscale)
    axes.set_yscale(yscale)
    axes.set_xlim(xlim)
    axes.set_ylim(ylim)
    if legend:
        axes.legend(legend)
    axes.grid()
```
