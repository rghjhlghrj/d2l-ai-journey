```python
# 【1】导入 torch。torch 是 PyTorch 的核心库，张量操作都在这里
import torch
```

```python
# 【2】arange 生成 0..11 的连续整数向量，view 改变形状为 3行4列 的矩阵
x = torch.arange(12)
x
```

```python
# 【3】用嵌套列表直接创建张量（张量中的元素类型必须一致）
x.shape
```

```python
# 【4】按位置索引访问：用 [-1] 取最后一行，[1:3] 取第2~3行（左闭右开）
x.numel()
```

```python
# 【5】逐元素运算：+ - * / 和 ** 幂运算，都是对应位置独立计算
X = x.reshape(3, 4)
X
```

```python
# 【6】一元运算：exp 是按元素求指数（e^x）
torch.zeros((2, 3, 4))
```

```python
# 【7】张量拼接（合并）：dim=0 沿行堆叠，dim=1 沿列并排
torch.ones((2, 3, 4))
```

```python
# 【8】逻辑运算：X == Y 生成同形状的布尔张量（逐位置比较）
torch.randn(3, 4)
```

```python
# 【9】sum() 求所有元素的和，得到标量（0维张量）
torch.tensor([[2, 1, 4, 3], [1, 2, 3, 4], [4, 3, 2, 1]])
```

```python
# 【10】广播机制：a是3x1，b是1x2，运算时各自'复制扩展'成3x2再相加
x = torch.tensor([1.0, 2, 4, 8])
y = torch.tensor([2, 2, 2, 2])
x + y, x - y, x * y, x / y, x ** y  # **运算符是求幂运算
```

```python
# 【11】id() 验证：Y = X + Y 后 id 变了，说明是新建对象；后面会学原地操作节省内存
torch.exp(x)
```

```python
# 【12】转成 NumPy 数组与反向转换：两者共享底层内存，转换零拷贝
X = torch.arange(12, dtype=torch.float32).reshape((3,4))
Y = torch.tensor([[2.0, 1, 4, 3], [1, 2, 3, 4], [4, 3, 2, 1]])
torch.cat((X, Y), dim=0), torch.cat((X, Y), dim=1)
```
